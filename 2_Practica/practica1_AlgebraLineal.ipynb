### 🎯 Ejercicio 1 — Álgebra Lineal Aplicada al Mundo Real

**Contexto:** Estás trabajando como Data Scientist en un banco. Tu tarea es construir un sistema de *credit scoring* — un número del 0 al 100 que indica qué tan confiable es un cliente para pagar un préstamo.

Los datos que tienes de cada cliente son:
- **ingreso_normalizado:** Qué tan alto es su ingreso (0 a 1)
- **historial_credito:** Qué tan buena es su historia de pagos (0 a 1)
- **ratio_deuda:** Cuánta deuda tiene respecto a sus ingresos (0 a 1, más alto = peor)
- **antigüedad:** Cuánto tiempo lleva como cliente (0 a 1)

El modelo ya aprendió unos **pesos** (importancia de cada variable). Tu trabajo es:
1. Calcular el score de cada cliente con álgebra lineal (`clientes @ pesos`)
2. Normalizar los scores al rango 0-100
3. Tomar decisiones de aprobación/rechazo

**¿Por qué álgebra lineal y no un for loop?**

```python
# Con un bucle (lento, feo):
scores = []
for cliente in clientes:
    score = 0
    for i, peso in enumerate(pesos):
        score += cliente[i] * peso
    scores.append(score)

# Con álgebra lineal (rápido, elegante):
scores = clientes @ pesos
```

En un banco real, puedes tener millones de clientes. La versión matricial es **cientos de veces más rápida**.

````python
import numpy as np
#Datos cambiados
"""--Score Crediticio con Algebra Lineal---

Datos de 5 clientes:
Columnas: [ingreso_norm, hist_credito, deuda_ratio, antiguedad]
"""
clientes = np.array([
    [0.82, 0.94, 0.18, 0.74],   # Cliente A
    [0.35, 0.63, 0.68, 0.24],   # Cliente B
    [0.58, 0.72, 0.38, 0.53],   # Cliente C
    [0.88, 0.52, 0.08, 0.91],   # Cliente D
    [0.25, 0.28, 0.93, 0.12],   # Cliente E
])

""" Pesos del modelo crediticio (aprendidos por ML, aquí dados manualmente)
 Cada peso indica la importancia de cada característica"""
pesos = np.array([0.33,   # ingreso → alta importancia positiva
                  0.32,   # hist_credito → alta importancia positiva
                  -0.22,  # deuda_ratio → penaliza el score
                  0.13])  # antiguedad → importancia moderada

""" 1) calcular scores con producto matricial"""
scores = np.dot(clientes, pesos)

print('--- Sistema de Score Crediticio ---')
print('                  Ingr  Hist  Deuda  Antig   SCORE')
nombres = ['A', 'B', 'C', 'D', 'E']
for nombre, datos, score in zip(nombres, clientes, scores):
    decision = 'APROBAR' if score >= 0.5 else 'RECHAZAR'
    print(f'  Cliente {nombre}: {datos}  →  {score:.3f}  {decision}')
print()

"""2) Normalizar scores al rango [0, 1]
   Fórmula: (x - min) / (max - min)"""
scores_min = scores.min()
scores_max = scores.max()
scores_norm = (scores - scores_min) / (scores_max - scores_min)

print('Scores normalizados a rango [0-100]:')
for nombre, sn in zip(nombres, scores_norm * 100):
    barra = '█' * int(sn / 5)
    print(f'  Cliente {nombre}: {sn:5.1f}/100  {barra}')
print()

"""---PASO 3: Estadísticas del portfolio---"""
print('Estadísticas del portfolio de clientes:')
print(f'  Score promedio:     {np.mean(scores_norm*100):.1f}')
print(f'  Score más alto:     {np.max(scores_norm*100):.1f} (Cliente {nombres[np.argmax(scores)]})')
print(f'  Score más bajo:     {np.min(scores_norm*100):.1f} (Cliente {nombres[np.argmin(scores)]})')
print(f'  Clientes aprobados: {np.sum(scores >= 0.5)}/{len(scores)}')
````
