## 🟢 Nivel Básico — Ejercicios 1 a 5

Estos ejercicios se resuelven con `SELECT`, `WHERE`, `ORDER BY`, `DISTINCT` y `LIMIT`.

### 🟢 Ejercicio 1 — Conocer el catálogo de productos

Muestra el **nombre**, la **categoría** y el **precio** de todos los productos,  
ordenados de **mayor a menor precio**.

**Resultado esperado:** 10 filas con las columnas nombre, categoria, precio.

```python
# ✏️ Escribe tu solución aquí
cursor.execute("""
SELECT nombre, categoria, precio
FROM productos
ORDER BY precio DESC
""")
for fila in cursor.fetchall():
    print(fila)
```

### 🟢 Ejercicio 2 — Clientes activos de Bogotá

Lista el **nombre** y la **edad** de todos los clientes que:  
- Viven en **Bogotá**, Y  
- Están **activos** (`activo = 1`)  

Ordena el resultado por **edad de mayor a menor**.

**Pista:** necesitas `WHERE` con `AND`.

```python
# ✏️ Escribe tu solución aquí
cursor.execute("""
SELECT nombre, edad
FROM clientes
WHERE ciudad = 'Bogotá' AND activo = 1
""")
for fila in cursor.fetchall():
    print(fila)
```

### 🟢 Ejercicio 3 — ¿Cuántas ciudades diferentes hay?

Obtén la **lista de ciudades únicas** donde tenemos clientes registrados.  
El resultado debe mostrar cada ciudad **una sola vez**, ordenado alfabéticamente.

**Pista:** usa `DISTINCT`.

```python
# ✏️ Escribe tu solución aquí
cursor.execute("""
SELECT DISTINCT ciudad 
FROM clientes
ORDER BY ciudad ASC
""")
for fila in cursor.fetchall():
    print(fila)
```

### 🟢 Ejercicio 4 — Productos económicos con stock disponible

Muestra el **nombre** y el **precio** de los productos que:  
- Tienen un precio **menor o igual a $300.000**, Y  
- Tienen **más de 40 unidades** en stock  

**Pista:** dos condiciones en el `WHERE` con `AND`.

```python
# ✏️ Escribe tu solución aquí
cursor.execute("""
SELECT nombre, Precio
FROM productos
WHERE precio <= 300000 AND stock > 40
""")
for fila in cursor.fetchall():
    print(fila)
```

### 🟢 Ejercicio 5 — Las 3 ventas más grandes

Muestra el **id_venta**, la **fecha** y el **total** de las **3 transacciones con mayor valor**.

## 🟡 Nivel Medio — Ejercicios 6 a 10

Aquí ya entra `GROUP BY`, `HAVING` y las funciones de agregación: `COUNT`, `SUM`, `AVG`, `MAX`, `MIN`.

### 🟡 Ejercicio 6 — Clientes por ciudad

Cuenta **cuántos clientes hay en cada ciudad**.  
Muestra la ciudad y el total de clientes, ordenado de **mayor a menor cantidad**.

**Pista:** `GROUP BY ciudad` + `COUNT(*)`.

```python
# ✏️ Escribe tu solución aquí
cursor.execute("""
SELECT ciudad,
       COUNT(*) AS numero_clientes
FROM clientes 
GROUP BY ciudad
""")
for fila in cursor.fetchall():
    print(fila)
```

### 🟡 Ejercicio 7 — Precio promedio por categoría de producto

Calcula el **precio promedio** y el **precio máximo** de los productos,  
agrupado por **categoría**.

Redondea el promedio a **0 decimales** con `ROUND()`.  
Ordena por precio promedio de mayor a menor.

**Pista:** `GROUP BY categoria` + `AVG()` + `MAX()`.

```python
# ✏️ Escribe tu solución aquí
cursor.execute("""
SELECT ROUND(AVG(precio), 0) AS precio_promedio,
       ROUND(MAX(precio), 0) AS precio_maximo
FROM productos
GROUP BY categoria
""")
for fila in cursor.fetchall():
    print(fila)
```

### 🟡 Ejercicio 8 — Ingresos totales por mes

Calcula el **total de ingresos** de cada mes del año 2024.  
Muestra el mes (formato `YYYY-MM`) y la suma de ventas, ordenado cronológicamente.

**Pista:** usa `SUBSTR(fecha, 1, 7)` para extraer el año y mes de la fecha.

```python
# ✏️ Escribe tu solución aquí
cursor.execute("""
SELECT SUBSTR(fecha, 1, 7) AS Fecha,
       SUM(total) AS total_ingresos
FROM ventas
WHERE fecha LIKE '2024-%'
GROUP BY SUBSTR(fecha, 1, 7)
ORDER BY SUBSTR(fecha, 1, 7) ASC
""")
for fila in cursor.fetchall():
    print(fila)
```

### 🟡 Ejercicio 9 — Clientes que han comprado más de una vez

Identifica los **id_cliente** que tienen **más de 1 venta registrada**.  
Muestra el id_cliente y la cantidad de compras que hizo.

**Pista:** `GROUP BY id_cliente` + `HAVING COUNT(...) > 1`.

```python
# ✏️ Escribe tu solución aquí
cursor.execute("""
SELECT id_cliente, 
    COUNT(*) AS cantidad_compras
FROM ventas
GROUP BY id_cliente
HAVING COUNT(*) > 1;
""")
for fila in cursor.fetchall():
    print(fila)
```

### 🟡 Ejercicio 10 — Segmentos con ticket promedio alto

Calcula el **ticket promedio** (valor promedio por venta) para cada **segmento de cliente** (`categoria`).  
Muestra solo los segmentos cuyo ticket promedio **supere los $500.000**.

**Pista:** haz JOIN entre `ventas` y `clientes`, agrupa por `categoria` y usa `HAVING AVG(...) > 500000`.

```python
# ✏️ Escribe tu solución aquí
cursor.execute("""
SELECT 
    clientes.categoria,
    AVG(ventas.total) AS ticket_promedio
FROM ventas
JOIN clientes
    ON ventas.id_cliente = clientes.id_cliente
GROUP BY clientes.categoria
HAVING AVG(ventas.total) > 500000;
""")
for fila in cursor.fetchall():
    print(fila)
```
